# Task, Workflow and Provenance Overview
A scientific workflow is a series of steps that involve data collection, manipulation, and analysis in order to produces some novel data product. The derivation of the product is thoroughly defined, as to augment the replicability of the workflow and consequently the trustworthiness of the data product. The entire process-- from raw data to useful data product-- is the ***workflow***. The individual steps that constructs the workflow are the ***tasks***. The workflow and tasks themselves are abstract-- they do not depend on any definite dataset as inputs, or predefine the values in any data product. Rather, they are the predefined procedures used to derive targeted data products from compatible data inputs.

When the workflow is enacted over some data, the actual record of the tasks being completed and the intermediate data products are known as the *provenance* of that data product.

## Tasks
A task represents an executable process that transforms data. It can be considered a function wrapper, where inputs are turned into outputs.

Tasks operate on a collection of inputs. These inputs can be other tasks' outputs, external data sources, or some other source-- such as a generated data source like noise.
Notice the directionality of the graph. The reverse flow represents a slice of the ***provenance*** relating to a data output. It traces the path of an output back to its raw origins. This is more clearly visible in the context of a full workflow.

## Workflows
A workflow is a defined chain of tasks, where the chaining is done by connecting the outputs of specific tasks to the input of another. An example workflow $W_1$ is given below. It is comprised of tasks in the form $T_n$, where $n$ identifies the $n$-th step of the workflow.

![Basic workflow diagram](basic_workflow.png)

In the above workflow, notice there are two types of tasks. The cylindrical tasks represent ***head tasks***, where an external datasource enters into the workflow. All workflows start with a head task-- as all workflows need some point for data entry. 
Pointing at the head tasks are internal tasks, or ***process tasks***. These tasks carry out the processing, manipulation, analysis, or some other action performed on the data to derive a final data product. 

# Provenance
The ***provenance*** of a data product is the pedigree of tasks that trace back to its original raw data source. The pedigree records the task, the actor, the data inputs and data outputs that relate to some end or intermediate product.

# Formal Definitions

A workflow, $\Omega$, is informally a defined set of tasks that process raw data inputs to targeted data products. More formally, a workflow is a directed acyclic graph ($DAG$), where vertices $V_\delta$ are intermediate data products, and edges $E_\lambda$ are algorithms that transform a source into its target.
$$\Omega = (V_\delta, E_\lambda)$$


The $DAG$ starts with at least one head vertex, notated $V_\Delta$. The head vertices are the entry points for raw data to enter a workflow. From this we can derive that any edge connecting two non-head vertices transform data that has already been transformed by a prior task. The edges connecting to a head vertex are denoted as $E_\Lambda$, and are responsible for translating raw data in an ambiguous form to a type compatible with depending tasks. 

The exhaustive set of tasks of a workflow, $\Gamma^*$, is comprised of all pairs of vertices that share an edge. An individual task, $\tau$, thus represents one algorithmic manipulation of an intermediate data input into an intermediate data product.
$$ \Gamma^* = \set{ \tau \in \Omega \ |\  \tau = (n, x, m),\  n, m \in V_\delta,\ x \in E_\lambda,\   \text{ and } n\  \overset{x}{\to}\  m} $$


The fundamental set of tasks of a workflow, $\Gamma$, is similar to the exhaustive set, $\Gamma^*$, but groups any tasks with the same $E_\lambda$ and $V_{\delta,\mathttt{out}}$ together. This set is more representative of computable tasks in a workflow, as a task may have multiple data inputs

Thus, a single task is defined by its triplet: 
$$ \tau = \begin{cases} \set{V_{\delta,\texttt{in}}}\\ E_{\lambda}\\ V_{\delta,\texttt{out}} \end{cases}$$

$$ \Gamma = \set{ \tau \in \Omega \ |\  \tau = (N, x, m),\  m \in V_\delta,\ x \in E_\lambda,\   \text{ and } N = \set{n \in V_\delta\ |\ n\ \overset{x}{\to}\  m}} $$
