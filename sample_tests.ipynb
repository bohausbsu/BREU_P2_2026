```python
import ipytest
ipytest.autoconfig()
```

```python
%%ipytest -qq

def test_example():
    assert [1, 2, 3] == [1, 2, 3]
```

```python
import Workflow, Tasks
def createTask(workflowID, taskID, dataInput):
    return Tasks.Task(workflowID, taskID, dataInput)

def createNextTask(task):
    return Tasks.Task(task.workflowID, task.taskID + 1, task)
```

```python
%%ipytest
def test_task_creation():
    wID = 12
    tID = 1
    task = createTask(wID, tID, list([]))
    assert task.workflowID == wID
    assert task.taskID == tID
    assert task.inputTaskUUIDs == []
```

```python
%%ipytest
def test_task_chaining():
    task1 = createTask(1, 1, list([]))
    task2 = createNextTask(task1)
    assert task2.workflowID == task1.workflowID
    assert task2.taskID == task1.taskID + 1
    assert task2.inputTaskUUIDs == [task1]
```
